## Project 2: JSON Config Editor


```python
from copy import deepcopy
from project_2_config_editor import load_config, update_config

config = load_config()
demo_config = deepcopy(config)
updated_config = update_config(demo_config, "font_size", "20")

print("Original config:")
print(json.dumps(config, indent=2, ensure_ascii=False))
print("\nDemo updated config:")
print(json.dumps(updated_config, indent=2, ensure_ascii=False))
```
