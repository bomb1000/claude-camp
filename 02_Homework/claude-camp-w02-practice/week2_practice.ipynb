## Exercise 1: Student Roster Manager
Use a dictionary to store student records and support add, search, and delete operations.

````python
roster = {}

def add_student(name, email, join_date):
    roster[name.lower()] = {"name": name, "email": email, "join_date": join_date}

def find_student(name):
    return roster.get(name.lower(), "Student not found.")

def delete_student(name):
    return roster.pop(name.lower(), "Student not found.")

add_student("Alice", "alice@example.com", "2026-05-16")
print(find_student("alice"))
print(delete_student("Alice"))
print(find_student("Alice"))
````

## Exercise 2: Word Frequency Counter
Count each word with a dictionary, ignore case, and sort by frequency.

````python
import string

def count_words(text):
    counts = {}
    for raw_word in text.split():
        word = raw_word.strip(string.punctuation).lower()
        if word:
            counts[word] = counts.get(word, 0) + 1
    return sorted(counts.items(), key=lambda item: (-item[1], item[0]))

for word, count in count_words("Hello hello Python, python world!"):
    print(f"{word}: {count}")
````

## Exercise 3: Todo List with File Saving
Store todo data in `todos.json`, load it safely, and handle missing files without crashing.

````python
import json
from pathlib import Path

path = Path("sample_todos.json")
def load_todos():
    try:
        return json.loads(path.read_text())
    except FileNotFoundError:
        return []

todos = load_todos()
todos.append({"task": "Finish homework", "done": False})
todos[0]["done"] = True
path.write_text(json.dumps(todos, indent=2))
print(json.loads(path.read_text()))
path.unlink()
````

## Exercise 4: Safe Calculator
Support arithmetic while handling invalid numbers, division by zero, and graceful exit.

````python
def calculate(left, operator, right):
    if operator == "+":
        return left + right
    if operator == "-":
        return left - right
    if operator in ("*", "x", "×"):
        return left * right
    if operator in ("/", "÷"):
        return "Cannot divide by zero." if right == 0 else left / right
    return "Unsupported operator."

print(calculate(10, "/", 0))
print(calculate(5, "+", 7))
print(calculate(5, "?", 7))
````
