## Exercise 1: Temperature Converter
Convert Celsius to Fahrenheit and Fahrenheit to Celsius.

```python
def celsius_to_fahrenheit(celsius):
    return celsius * 9 / 5 + 32

def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5 / 9

print(f"25.0 C = {celsius_to_fahrenheit(25):.1f} F")
print(f"77.0 F = {fahrenheit_to_celsius(77):.1f} C")
```

## Exercise 2: Personalized Greeting
Input a name and age, then output a formatted greeting.

```python
def build_greeting(name, age):
    return (
        f"Hello, {name}! You are {age} years old.\n"
        f"Next year, you will be {age + 1}.\n"
        "Keep learning and building something useful every week!"
    )

print(build_greeting("Michael", 25))
```

## Exercise 3: Tip Calculator
Input bill amount and tip percentage, then output the tip and total amount.

```python
def calculate_tip(bill_amount, tip_percent):
    tip = bill_amount * tip_percent / 100
    total = bill_amount + tip
    return tip, total

tip, total = calculate_tip(100, 15)
print(f"Bill amount: $100.00")
print(f"Tip (15.0%): ${tip:.2f}")
print(f"Total amount: ${total:.2f}")
```

## Exercise 4: Simple Password Generator
Input password length, then generate a random string password.

```python
import random
import string

def generate_password(length):
    random.seed(7)
    characters = string.ascii_letters + string.digits + string.punctuation
    return "".join(random.choice(characters) for _ in range(length))

print(f"Generated password: {generate_password(12)}")
```

## Exercise 5: BMI Calculator
Input height and weight, then output BMI value and health advice.

```python
def calculate_bmi(weight_kg, height_cm):
    height_m = height_cm / 100
    return weight_kg / (height_m ** 2)

def get_advice(bmi):
    if bmi < 18.5:
        return "Underweight: consider balanced meals and strength training."
    if bmi < 24:
        return "Healthy range: keep your current habits."
    if bmi < 27:
        return "Overweight: add regular exercise and watch portions."
    return "Obese range: consider speaking with a healthcare professional."

bmi = calculate_bmi(65, 170)
print(f"Your BMI is {bmi:.1f}")
print(get_advice(bmi))
```
